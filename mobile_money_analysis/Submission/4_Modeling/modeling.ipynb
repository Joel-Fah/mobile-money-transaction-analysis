## Load best model & make a prediction

```
bundle = joblib.load('best_model.pkl')
model, le, feats = bundle['model'], bundle['label_encoder'], bundle['features']
sample = df[feats].iloc[:3]
pred = le.inverse_transform(model.predict(sample))
pd.DataFrame({'UserId': df['UserId'].iloc[:3], 'Predicted': pred})
```
